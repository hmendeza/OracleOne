## Most and Least Sold Products for Tienda 1

### Subtask:
Calculate and visualize the top 5 most sold and bottom 5 least sold products for 'tienda'.


**Reasoning**:
To visualize the most and least sold products for 'tienda', I will first group the DataFrame by the 'Producto' column and count the occurrences of each product. Then, I will extract the top 5 and bottom 5 products based on their sales counts. Finally, I will use `matplotlib.pyplot` to create two separate bar plots for these selections. I need to import `matplotlib.pyplot` for plotting.



```python
import matplotlib.pyplot as plt

# Calculate product sales frequency for tienda
product_sales = tienda['Producto'].value_counts()

# Identify top 5 most sold products
top_5_products = product_sales.head(5)

# Identify bottom 5 least sold products
bottom_5_products = product_sales.tail(5)

# Plotting Top 5 Most Sold Products
plt.figure(figsize=(10, 6))
top_5_products.plot(kind='bar', color='skyblue')
plt.title('Top 5 Most Sold Products in Tienda 1')
plt.xlabel('Product')
plt.ylabel('Number of Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Plotting Bottom 5 Least Sold Products
plt.figure(figsize=(10, 6))
bottom_5_products.plot(kind='bar', color='lightcoral')
plt.title('Bottom 5 Least Sold Products in Tienda 1')
plt.xlabel('Product')
plt.ylabel('Number of Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
